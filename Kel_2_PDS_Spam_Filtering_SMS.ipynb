# **PREPROCESSING DATA**

```python
# Membuat objek Stemmer menggunakan Sastrawi
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Fungsi untuk melakukan preprocessing teks
def transform_text(text):
    text = text.lower()  # Mengubah semua huruf menjadi huruf kecil
    text = re.sub(r'[^\w\s]', '', text)  # Menghapus tanda baca dari teks
    tokens = nltk.word_tokenize(text)  # Melakukan tokenisasi (memecah teks menjadi kata-kata)
    tokens = [word for word in tokens if word not in stopwords.words('indonesian')]  # Menghapus stopwords (kata umum yang tidak bermakna)
    tokens = [stemmer.stem(word) for word in tokens]  # Melakukan stemming (mengubah kata menjadi bentuk dasar)
    return " ".join(tokens) # Menggabungkan kembali token menjadi teks

# Menerapkan preprocessing pada kolom 'Message' di DataFrame
spam_df['processed_message'] = spam_df['Message'].apply(transform_text)
```

```python
# Menampilkan perbandingan data sebelum dan setelah preprocessing
print("Data PDS Kel 2.csv sebelum preprocessing:")
print(spam_df)

print("\n Data PDS Kel 2.csv setelah preprocessing:")
print(spam_df['processed_message'])
```

```python
# Mengatur agar teks ditampilkan secara penuh
pd.set_option('display.max_colwidth', None)

# Menampilkan data setelah preprocessing
# Misalkan data Anda disimpan di DataFrame bernama `spam_df` dan kolom hasil preprocessing adalah 'processed_message'

print("Data setelah preprocessing (tampilan lengkap):")
print(spam_df[['Message', 'processed_message']])
```

```python
# Membagi dataset menjadi data latih (training) 80% dan data uji (testing) 20%
x_train, x_test, y_train, y_test = train_test_split(
    spam_df['processed_message'], spam_df['Label'],
    test_size=0.2, random_state=42, stratify=spam_df['Label']
)
```

# **PEMBOBOTAN DENGAN TF-IDF**

```python
# Melakukan vektorisasi teks dengan model TF-IDF (Term Frequency-Inverse Document Frequency)
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))
# Mengonversi data teks pada x_train ke dalam bentuk vektor numerik
x_train_tfidf = tfidf.fit_transform(x_train)
# Mentransformasi data teks pada x_test menggunakan model TF-IDF yang telah dilatih
x_test_tfidf = tfidf.transform(x_test)
```

```python
# Menampilkan data yang sudah ditransformasi
print("\nData Training")
print(x_train_tfidf)
print("\nData Testing")
print(x_test_tfidf)
```

# **KLASIFIKASI NAIVE BAYES**

```python
# Membuat model klasifikasi menggunakan Naive Bayes (Multinomial Naive Bayes)
model = MultinomialNB()
# Melatih model dengan data latih (fit)
model.fit(x_train_tfidf, y_train)
```

# **EVALUASI**

```python
# Melakukan prediksi pada data uji menggunakan model yang telah dilatih
y_pred = model.predict(x_test_tfidf)

# Mengevaluasi performa model
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred)) # Menampilkan Confusion Matrix
print("\nClassification Report:")
print(classification_report(y_test, y_pred)) # Menampilkan Classification Report (precision, recall, f1-score, support)
print("\nAccuracy Score:", accuracy_score(y_test, y_pred)) # Menampilkan skor akurasi model
```

```python
# Menguji model dengan input manual (teks baru)
test_sms_ham = ["Halo Dila"] # Contoh teks manual yang diasumsikan sebagai 'ham'
test_sms_spam = ["Ada Paket Ekslusif"] # Contoh teks manual yang diasumsikan sebagai 'spam'

# Melakukan transformasi teks manual menggunakan model TF-IDF yang telah dilatih
test_sms_ham_tfidf = tfidf.transform(test_sms_ham)
test_sms_spam_tfidf = tfidf.transform(test_sms_spam)

# Melakukan prediksi kategori menggunakan model yang telah dilatih
print("Ham Test Prediction:", model.predict(test_sms_ham_tfidf))
print("Spam Test Prediction:", model.predict(test_sms_spam_tfidf))
```

# **CADANGAN SINTAKS AKA SINTAKS ASLIII**

```python
factory = StemmerFactory()
stemmer = factory.create_stemmer()

def transform_text(text):
    text = text.lower()  # Lowercase
    text = re.sub(r'[^\w\s]', '', text)  # Remove punctuation
    tokens = nltk.word_tokenize(text)  # Tokenize
    tokens = [word for word in tokens if word not in stopwords.words('indonesian')]  # Remove stopwords
    tokens = [stemmer.stem(word) for word in tokens]  # Stemming
    return " ".join(tokens)

spam_df['processed_message'] = spam_df['Message'].apply(transform_text)

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(
    spam_df['processed_message'], spam_df['Label'],
    test_size=0.2, random_state=42, stratify=spam_df['Label']
)

# TF-IDF Vectorization
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))
x_train_tfidf = tfidf.fit_transform(x_train)
x_test_tfidf = tfidf.transform(x_test)

# Train Naive Bayes classifier
model = MultinomialNB()
model.fit(x_train_tfidf, y_train)

# Predict on test data
y_pred = model.predict(x_test_tfidf)

# Evaluate the model
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
print("\nAccuracy Score:", accuracy_score(y_test, y_pred))
```

```python
# Testing model with manual input
test_sms_ham = ["Halo Dila"]
test_sms_spam = ["Ada Paket Ekslusif"]

test_sms_ham_tfidf = tfidf.transform(test_sms_ham)
test_sms_spam_tfidf = tfidf.transform(test_sms_spam)

print("Ham Test Prediction:", model.predict(test_sms_ham_tfidf))
print("Spam Test Prediction:", model.predict(test_sms_spam_tfidf))
```
